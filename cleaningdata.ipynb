STEP2 - DATA QUALITY ASSESMENT 

```python
#Missing values
missing_report=pd.DataFrame({
    "Missing count": data.isna().sum(),
    "Missing percentage":(data.isna().sum()/len(data))*100
})

print("Missing value report")
print(missing_report)
```

```python
#Duplicates 
print("Duplicate Values")
print(data.duplicated().sum())
```

```python
#Minimum values
print("Minimum Values\n")
print(data.select_dtypes(include=np.number).min())
```

```python
#Maximum number 
print("Maximum Numbers\n")
print(data.select_dtypes(include=np.number).max())
```

```python
#Detecting outlier using IQR (Inter Quartile Range)=Q3-Q1
#Select numerical columns
num_columns=data.select_dtypes(include=np.number).columns

outlier_report={}
 
for column in num_columns:
    Q1=data[column].quantile(0.25) #Quartile 1st of numerical columns
    Q3=data[column].quantile(0.75) #Quartile 3rd of numerical columns

    IQR=Q3-Q1

    lower_bound=Q1-1.5* IQR
    upper_bound=Q3+1.5*IQR

    outlier=data[
        (data[column]<lower_bound) |
        (data[column]>upper_bound)
    ]

    outlier_report[column]=len(outlier)
    
```

```python
#Displaying Outlier Report
outlier_df=pd.DataFrame(outlier_report.items(),
columns=["Columns","Outlier_count"])

print("Outlier Detection Report")
print(outlier_df)
```

```python
#Convert TimeStamp to date time 
data['timestamp']=pd.to_datetime(
    data["timestamp"],errors="coerce"
)

print("TimeStamp DataTypes ")
print(data['timestamp'].dtype)
```

```python
#Checking Invalid Time stamp report 
print("Invalid Timestamps:")
print(data["timestamp"].isnull().sum())
```

STEP 3- CLEANING DATA AND OUTLIER CAPPING


```python
#Create a copy of original data 
original_data=data.copy()
```

```python
#Verfying is their any missing value or not 
missing_values=original_data.isna().sum().sum()
if missing_values==0:
    print("No null values are their")
else:
    print("Null values are Present ",missing_values)
```

```python
#Verifying is their duplicate values or not 
duplicate_Values=original_data.duplicated().sum().sum()
if duplicate_Values==0:
    print("No Duplicate values are Present")
elif duplicate_values>0:
    original_data.drop_duplicates(inplace=True)
    print("Duplicate records deleted ")
else:
    print("Duplicate values prsent",duplicate_values)
```

```python
#Handle Outlier Using IQR Capping
outlier_columns=['fuel_consumption_rate','iot_temperature','disruption_likelihood_score']

for column in outlier_columns:
    Q1=original_data[column].quantile(0.25)
    Q3=original_data[column].quantile(0.75)

    IQR=Q3-Q1
    lower_bound=Q1-1.5*IQR
    upper_bound=Q3+1.5*IQR

    original_data[column]=original_data[column].clip(
        lower=lower_bound,upper=upper_bound
    )

    print(f"Outlier Handled in {column}")
    print("Lower Bound ",lower_bound)
    print("Upper Bound",upper_bound)

    
```

```python
#Final Verification 
print("Final Dataset Shape:", original_data.shape)
print("Missing Values:", original_data.isnull().sum().sum())
print("Duplicates:", original_data.duplicated().sum())
```

```python
#Saving cleaned dataset
original_data.to_csv("Cleaned_logistic_dataset.csv",index=False)
```

STEP4- Data Normalization

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
#Create copy of cleaned dataset
normalize_Data=original_data.copy()
#Selecting numerical columns for normalization
num_Columns=normalize_Data.select_dtypes(include=np.number).columns
#Initialize MinMaxScaler
scaler=MinMaxScaler()
```

```python
#Applying normalisation
normalize_Data[num_Columns]=scaler.fit_transform(normalize_Data[num_Columns])
```

```python
#Saving Normalized data 
normalize_Data.to_csv("normalized_logistic_Data.csv",index=False)
print("Data Save successfulll")
```

STEP 5- DATA VISUALIZATION

```python
#Outlier Comparision befor and after capping 
outlier_columns=['fuel_consumption_rate','iot_temperature','disruption_likelihood_score']
outlier_counts=[]
for column in outlier_columns:
    Q1=data[column].quantile(0.25)
    Q3=data[column].quantile(0.75)
    IQR=Q3-Q1
    lower_bound=Q1-1.5*IQR
    upper_bound=Q3+1.5*IQR

    before=(
        (data[column]<lower_bound)|
        (data[column]>upper_bound)
    ).sum()

    after=(
        (original_data[column]<lower_bound)|
        (original_data[column]>upper_bound)
    ).sum()

    outlier_counts.append({
        "Colums":column,
        "Before Capping":before,
        "After Capping":after
    })
```

```python
outlier_Data=pd.DataFrame(outlier_counts)
```

```python
plt.figure(figsize=(10, 6))

plot_data = outlier_Data.set_index("Colums")

ax = plot_data[
    ["Before Capping", "After Capping"]
].plot(kind="bar", figsize=(10, 6))

plt.title("Outlier Reduction After IQR Capping")
plt.xlabel("Numerical Features")
plt.ylabel("Number of Outliers")
plt.xticks(rotation=20)
plt.legend(title="Treatment")

# Display numerical values above bars
for container in ax.containers:
    ax.bar_label(container, fmt="%d", padding=3)

plt.ylim(0, plot_data["Before Capping"].max() * 1.15)

plt.tight_layout()
plt.savefig("outlier_comparison.png")
plt.show()
```

```python
#Distribution of Shipping cost  Befor and after Normalisation
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
sns.histplot(data['shipping_costs'],kde=True)
plt.title("Original Shipping Costs")

plt.subplot(1,2,2)
sns.histplot(normalize_Data['shipping_costs'],kde=True)
plt.title("Min-Max scaled Shipping Costs")
plt.tight_layout()
plt.savefig("shipping_costs_normalization.png")
plt.show()
```

```python
# 3. Boxplot of Normalized Numerical Features

selected_columns = [
    "fuel_consumption_rate",
    "shipping_costs",
    "lead_time_days",
    "iot_temperature"
]

plt.figure(figsize=(10, 6))

sns.boxplot(data=normalize_Data[selected_columns])

plt.title("Boxplot of Normalized Logistics Features")
plt.xticks(rotation=30)
plt.tight_layout()
plt.savefig("normalized_features_boxplot.png")
plt.show()

print("All visualizations generated successfully!")
```
